# TEORÍA DE CONJUNTOS

Cada píxel se convierte en una coordenada del elemento de la imagen y así podemos ver los objetos como:
$$ A, B \in \mathbb{R}^2 $$
Dónde cada píxel viene dado por las coordenadas:
$$ a = (a_1, a_2) $$

# Operaciones de conjuntos:
- Traslación:
$$ x : A + x = (A)_x = \{c|a+x, a \in A\} $$
- Reflexión:
$$ B^{-} = \{x|x = -b, b \in B\} = (-B) $$
- Complemento de $A$:
$$ A:A^c = \{x|x \notin A\} $$
- Diferencia:
$$ A - B = A \cap B^c = \{x | x \in A, x \notin B\} $$
